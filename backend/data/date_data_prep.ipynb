```python
import os
import json
from tqdm.notebook import tqdm  # Progress bar in Jupyter Notebook
```

```python
date_folder = r"D:\Project\NEPSE-Navigator\backend\data\date"  # Folder with raw date-wise JSON files
cleaned_folder = r"D:\Project\NEPSE-Navigator\backend\data\cleaned_date"  # Output folder for cleaned data

# Ensure cleaned folder exists
os.makedirs(cleaned_folder, exist_ok=True)
```

```python
sample_file = os.path.join(date_folder, os.listdir(date_folder)[0])  # Pick first file

with open(sample_file, "r", encoding="utf-8") as f:
    data = json.load(f)

# Print first few dates
print(json.dumps({k: data[k] for k in list(data.keys())[:3]}, indent=4))
```

```python
missing_values = 0

for filename in os.listdir(date_folder):
    if filename.endswith(".json"):
        file_path = os.path.join(date_folder, filename)

        with open(file_path, "r", encoding="utf-8") as f:
            data = json.load(f)

        for date, entry in data.items():
            if not isinstance(entry, dict):  # Ensure entry is a dictionary
                continue  

            if "metadata" in entry and isinstance(entry["metadata"], dict) and entry["metadata"].get("totalAmt", 0) == 0:
                continue  # Skip holidays (already no data)

            if "data" in entry and isinstance(entry["data"], list):  # Ensure 'data' is a list
                for stock in entry["data"]:
                    if isinstance(stock, dict) and "price" in stock and "diff" in stock["price"]:
                        if stock["price"]["diff"] is None:
                            missing_values += 1

print(f"⚠️ Found {missing_values} missing `diff` values.")
```
