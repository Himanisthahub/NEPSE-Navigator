```python
import os
import json
from tqdm.notebook import tqdm  # Use tqdm.notebook for better progress bars in Jupyter
```

```python
company_folder = "..data/company/"
```

```python
def clean_company_data():
    for filename in tqdm(os.listdir(company_folder), desc="Cleaning JSON Files"):
        if filename.endswith(".json"):
            file_path = os.path.join(company_folder, filename)
            
            with open(file_path, "r", encoding="utf-8") as f:
                data = json.load(f)

            # Fix missing "diff" values and sort by date
            cleaned_data = {}
            for date in sorted(data.keys()):  # Sort dates
                entry = data[date]
                if "price" in entry and "diff" in entry["price"]:
                    if entry["price"]["diff"] is None:
                        entry["price"]["diff"] = 0  # Replace null with 0
                
                cleaned_data[date] = entry
            
            # Save cleaned data back
            with open(file_path, "w", encoding="utf-8") as f:
                json.dump(cleaned_data, f, indent=4)

    print("✅ Company data cleaned successfully!")
```

```python
clean_company_data()
```
