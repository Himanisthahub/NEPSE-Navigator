```python
missing_values = 0

for filename in os.listdir(company_folder):
    if filename.endswith(".json"):
        file_path = os.path.join(company_folder, filename)

        with open(file_path, "r", encoding="utf-8") as f:
            data = json.load(f)

        for date, entry in data.items():
            if "price" in entry and "diff" in entry["price"]:
                if entry["price"]["diff"] is None:
                    missing_values += 1

print(f"⚠️ Found {missing_values} missing `diff` values.")
```

```python
duplicate_count = 0

for filename in os.listdir(company_folder):
    if filename.endswith(".json"):
        file_path = os.path.join(company_folder, filename)

        with open(file_path, "r", encoding="utf-8") as f:
            data = json.load(f)

        date_list = list(data.keys())
        if len(date_list) != len(set(date_list)):  # Compare list length before/after removing duplicates
            duplicate_count += 1

print(f"🔍 Duplicate dates found in {duplicate_count} files.")
```

```python
cleaned_folder = "data/cleaned_companydata/"

# Ensure cleaned folder exists
os.makedirs(cleaned_folder, exist_ok=True)

def clean_company_data():
    for filename in tqdm(os.listdir(company_folder), desc="Cleaning JSON Files"):
        if filename.endswith(".json"):
            input_file_path = os.path.join(company_folder, filename)
            output_file_path = os.path.join(cleaned_folder, filename)  # Save cleaned file in new folder
            
            with open(input_file_path, "r", encoding="utf-8") as f:
                data = json.load(f)

            cleaned_data = {}
            for date in sorted(data.keys()):  # Sort by date
                entry = data[date]
                if "price" in entry and "diff" in entry["price"]:
                    if entry["price"]["diff"] is None:
                        entry["price"]["diff"] = 0  # Replace null with 0
                
                cleaned_data[date] = entry
            
            # Save cleaned data to the new folder
            with open(output_file_path, "w", encoding="utf-8") as f:
                json.dump(cleaned_data, f, indent=4)

    print(f"✅ Cleaned data saved in '{cleaned_folder}' without modifying original files!")

clean_company_data()
```

```python
import os
import json
from tqdm import tqdm
from langchain.schema import Document
from langchain.vectorstores import FAISS
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.text_splitter import CharacterTextSplitter

stock_folder = "../data/cleaned_companydata/"  # Change this to your stock data folder

def load_and_chunk_stock_data(folder_path, chunk_size=1000, chunk_overlap=200):
    all_chunks = []
    file_count = 0
    for filename in tqdm(os.listdir(folder_path), desc="Processing Stock JSON Files"):
        if filename.endswith(".json"):
            file_path = os.path.join(folder_path, filename)
            file_count += 1
            
            with open(file_path, "r", encoding="utf-8") as f:
                data = json.load(f)
            
            # Print the first 5 data entries
            print(f"First 5 data entries in {filename}:")
            for i, (date, entry) in enumerate(data.items()):
                if i >= 5:
                    break
                print(f"{date}: {entry}")

            # Join all cleaned pages into one string
            cleaned_text = "\n".join(str(data[date].get("content", "")) for date in sorted(data.keys()) if "content" in data[date] and data[date]["content"])

            # Set up the text splitter
            text_splitter = CharacterTextSplitter(
                chunk_size=chunk_size,
                chunk_overlap=chunk_overlap,
                separator="."
            )

            # Split the joined cleaned text into chunks
            chunks = text_splitter.split_text(cleaned_text)
            all_chunks.extend(chunks)
    
    print(f"Total files processed: {file_count}")
    print(f"Total chunks created: {len(all_chunks)}")
    if all_chunks:
        print("Sample chunk content:")
        print(all_chunks[0])
    
    return [Document(page_content=chunk) for chunk in all_chunks]  # Convert to LangChain Document format

print("Loading and chunking stock data...")
stock_documents = load_and_chunk_stock_data(stock_folder)

# Ensure stock_documents is not empty
if stock_documents:
    # Define the embedding model
    embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

    # Store vectors using LangChain's FAISS wrapper
    stock_vector_db = FAISS.from_documents(stock_documents, embedding_model)

    # Save FAISS index locally
    stock_vector_db.save_local("final_ready_vector_db_data")
    print("Stock vector database saved successfully!")
else:
    print("Error: stock_documents list is empty. Please check the load_and_chunk_stock_data function.")
```

```python
import os
import json
from tqdm import tqdm
from langchain.schema import Document
from langchain.vectorstores import FAISS
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.text_splitter import CharacterTextSplitter

stock_folder = "../data/cleaned_companydata/"  # Change this to your stock data folder

def load_and_chunk_stock_data(folder_path, chunk_size=1000, chunk_overlap=200):
    all_chunks = []
    file_count = 0
    
    for filename in tqdm(os.listdir(folder_path), desc="Processing Stock JSON Files"):
        if filename.endswith(".json"):
            file_path = os.path.join(folder_path, filename)
            file_count += 1
            
            with open(file_path, "r", encoding="utf-8") as f:
                data = json.load(f)
            
            # Print the first 5 data entries for debugging
            print(f"\n🔍 First 5 data entries in {filename}:")
            for i, (date, entry) in enumerate(sorted(data.items())[:5]):
                print(f"{date}: {entry}")

            # ✅ Convert stock data to readable text format
            stock_text = "\n".join(
                f"Date: {date}, Open: {info['price'].get('open', 'N/A')}, "
                f"High: {info['price'].get('max', 'N/A')}, Low: {info['price'].get('min', 'N/A')}, "
                f"Close: {info['price'].get('close', 'N/A')}, Volume: {info.get('tradedShares', 'N/A')}, "
                f"Transactions: {info.get('numTrans', 'N/A')}, Amount: {info.get('amount', 'N/A')}"
                for date, info in sorted(data.items()) if 'price' in info
            )

            # ✅ Set up the text splitter
            text_splitter = CharacterTextSplitter(
                chunk_size=chunk_size,
                chunk_overlap=chunk_overlap,
                separator="\n"
            )

            # ✅ Split stock data into chunks
            chunks = text_splitter.split_text(stock_text)
            all_chunks.extend(chunks)
    
    print(f"\n📊 Total files processed: {file_count}")
    print(f"📌 Total chunks created: {len(all_chunks)}")
    
    # ✅ Print a sample chunk for debugging
    if all_chunks:
        print("\n📝 Sample chunk content:\n")
        print(all_chunks[0])
    
    return [Document(page_content=chunk) for chunk in all_chunks]  # Convert to LangChain Document format

print("🔄 Loading and chunking stock data...")
stock_documents = load_and_chunk_stock_data(stock_folder)

# ✅ Ensure stock_documents is not empty
if stock_documents:
    # ✅ Define the embedding model
    embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

    # ✅ Store vectors using LangChain's FAISS wrapper
    stock_vector_db = FAISS.from_documents(stock_documents, embedding_model)

    # ✅ Save FAISS index locally
    stock_vector_db.save_local("final_ready_vector_db_data")
    print("✅ Stock vector database saved successfully!")
else:
    print("❌ Error: `stock_documents` is empty. Please check the function.")
```
